```python
import os
import json

# Path to search
root_dir = "./ICON_sfc_ScenarioMIP.zarr_rechunked"
output_file = f"{root_dir}/.zmetadata"

# Container for all zarray contents
zmetadata = {"metadata": {},
             "zarr_consolidated_format": 1}

for dirpath, _, filenames in os.walk(root_dir):
    for filename in filenames:
        if filename.endswith(".zarray") or \
            filename.endswith(".zattrs") or \
            filename.endswith(".zgroup"):
            
            keys = dirpath.split("/")[2:]
            print(keys)
            if len(keys) < 1:
                arrkey = f"{filename}"
            else:
                keypath = "/".join(keys)
                arrkey = f"{keypath}/{filename}"
                
            file_path = os.path.join(dirpath, filename)
            try:
                with open(file_path, "r") as f:
                    data = json.load(f)
                    zmetadata["metadata"][arrkey] = data
            except Exception as e:
                print(f"Error reading {file_path}: {e}")
print(zmetadata)
```

```python
# Write combined JSON
with open(output_file, "w") as f:
    json.dump(zmetadata, f, indent=2)
```
